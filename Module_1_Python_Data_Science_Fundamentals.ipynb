```python
def add_numbers(a, b):
    result = a + b
    return result

answer = add_numbers(10, 20)
print("Sum:", answer)
```

```python
def check_number(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_number(10))
print(check_number(7))
```

```python
def calculate_average(a, b, c):
    average = (a + b + c) / 3
    return average

result = calculate_average(10, 20, 30)
print("Average:", result)
```

```python
import pandas as pd

data = {
    "Name": ["Abhishek", "Rahul", "Anil", "Priya"],
    "Age": [20, 21, 22, 20],
    "Marks": [85, 78, 92, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.head())
```

```python
print(df["Name"])
print(df["Marks"])
```

```python
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
```

```python
df["Pass"] = df["Marks"] >= 40

print(df)
```
